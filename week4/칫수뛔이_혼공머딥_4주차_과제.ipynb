```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

wine = pd.read_csv('https://bit.ly/wine_csv_data')

data = wine[['alcohol', 'sugar', 'pH']].to_numpy()
target = wine['class'].to_numpy()

train_input, test_input, train_target, test_target = train_test_split(data, target, test_size=0.2, random_state=42)
```

## RandomForestClassifier

```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint, uniform
import numpy as np

rf = RandomForestClassifier(random_state=42, n_jobs=-1)

param_dist = {
    'n_estimators': randint(100, 600),
    'max_depth': randint(3, 30),
    'min_samples_split': randint(2, 30),
    'min_samples_leaf': randint(1, 20),
    'max_features': ['sqrt', 'log2', None],
    'bootstrap': [True, False]
}

rf_search = RandomizedSearchCV(
    rf,
    param_distributions=param_dist,
    n_iter=40,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    random_state=42,
    verbose=0,
    return_train_score=True
)

rf_search.fit(train_input, train_target)

print("Best CV accuracy: {:.4f}".format(rf_search.best_score_))
print("Best params:", rf_search.best_params_)

best_rf = rf_search.best_estimator_
print("Train acc:", best_rf.score(train_input, train_target))
print("Test  acc:", best_rf.score(test_input, test_target))
```

## ExtraTreesClassifier


```python
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint, uniform

et = ExtraTreesClassifier(random_state=42, n_jobs=-1)

et_param_dist = {
    "n_estimators": randint(200, 1000),
    "max_depth": randint(3, 50),
    "min_samples_split": randint(2, 50),
    "min_samples_leaf": randint(1, 30),
    "max_features": ["sqrt", "log2", None],
    "bootstrap": [False, True],
    "criterion": ["gini", "entropy"],
}

et_search = RandomizedSearchCV(
    et,
    param_distributions=et_param_dist,
    n_iter=50, cv=5, scoring="accuracy",
    n_jobs=-1, random_state=42, return_train_score=True,
)

et_search.fit(train_input, train_target)

print("ET best CV acc:", round(et_search.best_score_, 4))
print("ET best params:", et_search.best_params_)

best_et = et_search.best_estimator_
print("ET train acc:", round(best_et.score(train_input, train_target), 4))
print("ET test  acc:", round(best_et.score(test_input, test_target), 4))
```

## GradientBoostingClassifier

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint, uniform

gb = GradientBoostingClassifier(random_state=42)

gb_param_dist = {
    "n_estimators": randint(100, 700),
    "learning_rate": uniform(0.01, 0.39),
    "max_depth": randint(2, 8),
    "min_samples_split": randint(2, 40),
    "min_samples_leaf": randint(1, 30),
    "subsample": uniform(0.5, 0.5),
    "max_features": ["sqrt", "log2", None],
}

gb_search = RandomizedSearchCV(
    gb,
    param_distributions=gb_param_dist,
    n_iter=50, cv=5, scoring="accuracy",
    n_jobs=-1, random_state=42, return_train_score=True,
)

gb_search.fit(train_input, train_target)

print("GBC best CV acc:", round(gb_search.best_score_, 4))
print("GBC best params:", gb_search.best_params_)

best_gbc = gb_search.best_estimator_
print("GBC train acc:", round(best_gbc.score(train_input, train_target), 4))
print("GBC test  acc:", round(best_gbc.score(test_input, test_target), 4))
```

## HistGradientBoostingClassifier

```python

from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint, uniform
import numpy as np

hgb = HistGradientBoostingClassifier(random_state=42)

param_dist = {
    'max_depth': randint(3, 40),
    'learning_rate': uniform(0.01, 0.39),
    'max_iter': randint(100, 800),
    'l2_regularization': uniform(0.0, 1.0),
    'min_samples_leaf': randint(5, 50)
}

hgb_search = RandomizedSearchCV(
    hgb,
    param_distributions=param_dist,
    n_iter=40,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    random_state=42,
    verbose=0,
    return_train_score=True
)

hgb_search.fit(train_input, train_target)

print("Best CV accuracy: {:.4f}".format(hgb_search.best_score_))
print("Best params:", hgb_search.best_params_)

best_hgb = hgb_search.best_estimator_
print("Train acc:", best_hgb.score(train_input, train_target))
print("Test  acc:", best_hgb.score(test_input, test_target))
```
